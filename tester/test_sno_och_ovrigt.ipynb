## Formfaktorer för snö (EN 1991-1-3 tabell 5.2)

μ1 = 0,8 för α ≤ 30°, 0,8(60 − α)/30 upp till 60°. μ2 = 0,8 + 0,8α/30 för α ≤ 30°, 1,6 för 30° < α < 60°.

```python
kolla("μ1(0)", lambda: kb.snö.μ1(0), 0.8)
kolla("μ1(45)", lambda: kb.snö.μ1(45), 0.4)
kolla("μ1(60)", lambda: kb.snö.μ1(60), 0.0)
kolla("μ2(0)", lambda: kb.snö.μ2(0), 0.8)
kolla("μ2(15)", lambda: kb.snö.μ2(15), 1.2, känt_fel=True)
kolla("μ2(45)", lambda: kb.snö.μ2(45), 1.6, känt_fel=True)
kolla("μ5 kontinuerlig vid 22,5°", lambda: kb.snö.μ5(22.5), 1.76 - 1.1 * 22.5 / 37.5)
```

## Välvning för I-tvärsnitt

Dubbelsymmetriskt: K_w = t b³ h²/24, W_w = t b² h/6, Z_w = t b² h/4. Enkelsymmetriskt med lika flänsar ska ge samma W_w. Med olika flänsar ska W_w vara det minsta av 2K_w/(b_ö h_ö) och 2K_w/(b_u h_u).

```python
K_w, W_w, Z_w = kb.tvärsnitt.dubbel_I(300, 20, 10, 600)
kolla("dubbel_I K_w", lambda: K_w, 8.1e12)
kolla("dubbel_I W_w", lambda: W_w, 1.8e8)
kolla("dubbel_I Z_w", lambda: Z_w, 2.7e8)
kolla("Enkelt_I symmetriskt W_w", lambda: kb.tvärsnitt.Enkelt_I(300, 20, 10, 300, 20, 600)[1], 1.8e8)
kolla("Enkelt_I K_w", lambda: kb.tvärsnitt.Enkelt_I(300, 20, 10, 150, 20, 600)[2], 1.8e12)
kolla("Enkelt_I W_w olika flänsar", lambda: kb.tvärsnitt.Enkelt_I(300, 20, 10, 150, 20, 600)[1], 4.5e7, känt_fel=True)
```
