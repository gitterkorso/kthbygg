# Test: rörprofiler (KKR och VKR)

Kontrollerar `kb.KKR` och `kb.VKR`. Cirkulärt rör 168,3×8 jämförs med tabellvärden: A = 40,3 cm², M = 31,6 kg/m, I = 1297 cm⁴, W_el = 154 cm³, W_pl = 206 cm³.

Kör med **Run All**. Kerneln ska vara `.venv` i repot.

- ✔ kontrollen gick igenom
- ✘ känt fel: fel som redan är kända och inte rättade än
- ✘ FEL: nytt fel

```python
import math
import kthbygg as kb
from kontroll import kolla, kolla_fel, klass, utskrift, sammanfattning, nollställ

nollställ()
```

## KKR cirkel 168,3×8

Returnerar d, A_s, A, M, I, i, W_el, W_pl, I_t, C_t.

```python
d, A_s, A, M, I, i, W_el, W_pl, I_t, C_t = kb.KKR.cirkel(168.3, 8, print_output=False)
kolla("d", lambda: d, 152.3)
kolla("A [mm²]", lambda: A, 4028.78)
kolla("I [mm⁴]", lambda: I, 12.9727e6)
kolla("i [mm]", lambda: i, math.sqrt(12.9727e6 / 4028.78))
kolla("W_pl [mm³]", lambda: W_pl, 205739)
kolla("I_t = 2I", lambda: I_t, 25.9454e6)
kolla("W_el [mm³]", lambda: W_el, 154162, känt_fel=True)
kolla("C_t = 2W_el [mm³]", lambda: C_t, 308324, känt_fel=True)
kolla("A_s [m²/m]", lambda: A_s, 0.52873, känt_fel=True)
kolla("M [kg/m]", lambda: M, 31.626, känt_fel=True)
```

## VKR cirkel 168,3×8

Samma formler som för KKR.

```python
d, A_s, A, M, I, i, W_el, W_pl, I_t, C_t = kb.VKR.cirkel(168.3, 8, print_output=False)
kolla("VKR A [mm²]", lambda: A, 4028.78)
kolla("VKR I [mm⁴]", lambda: I, 12.9727e6)
kolla("VKR W_el [mm³]", lambda: W_el, 154162, känt_fel=True)
kolla("VKR A_s [m²/m]", lambda: A_s, 0.52873, känt_fel=True)
kolla("VKR M [kg/m]", lambda: M, 31.626, känt_fel=True)
```

## KKR ellips som cirkel

En ellips med H = B = D ska ge samma värden som cirkeln. I_t jämförs med 2I (tunnväggigt rör, 1 % tolerans).

Returnerar A, I_y, I_z, i_y, i_z, W_y_el, W_z_el, W_y_pl, W_z_pl, I_t, C_t, A_s, M.

```python
A, I_y, I_z, i_y, i_z, W_y_el, W_z_el, W_y_pl, W_z_pl, I_t, C_t, A_s, M = kb.KKR.ellips(168.3, 168.3, 8, print_output=False)
kolla("ellips A", lambda: A, 4028.78)
kolla("ellips I_y", lambda: I_y, 12.9727e6)
kolla("ellips W_pl", lambda: W_y_pl, 205739)
kolla("ellips W_el", lambda: W_y_el, 154162, känt_fel=True)
kolla("ellips A_s = πD", lambda: A_s, 0.52873, känt_fel=True)
kolla("ellips M", lambda: M, 31.626, känt_fel=True)
kolla("ellips I_t ≈ 2I", lambda: I_t, 25.9454e6, rel=0.01, känt_fel=True)
```

## KKR RHS 200×100×5

Arean kontrolleras mot standardens formel A = 2T(B + H − 2T) − (4 − π)(r_o² − r_i²) med r_o = 2T, r_i = T. Övriga storheter jämförs med tabellvärden i EN 10219-2 (avrundade, 0,5 % tolerans): I_y = 1460 cm⁴, I_z = 497 cm⁴, W_pl,y = 181 cm³, I_t = 1210 cm⁴.

```python
A, I_y, I_z, i_y, i_z, W_y_pl, W_z_pl, W_y_el, W_z_el, I_t, r_o, r_i, A_s, M = kb.KKR.RHS(200, 100, 5, print_output=False)
A_ref = 2 * 5 * (100 + 200 - 10) - (4 - math.pi) * (10**2 - 5**2)
kolla("RHS A", lambda: A, A_ref)
kolla("RHS M = 0,00785·A", lambda: M, 0.00785 * A_ref)
kolla("RHS I_y", lambda: I_y, 1460e4, rel=0.005)
kolla("RHS I_z", lambda: I_z, 497e4, rel=0.005)
kolla("RHS W_pl,y", lambda: W_y_pl, 181e3, rel=0.005)
kolla("RHS I_t", lambda: I_t, 1210e4, rel=0.005)
kolla("RHS W_y_el = 2I_y/H", lambda: W_y_el, 2 * I_y / 200)
kolla("RHS print_output=True skriver ut", lambda: utskrift(kb.KKR.RHS, 200, 100, 5, print_output=True) != "", True, känt_fel=True)
kolla_fel("RHS ogiltig geometri", lambda: kb.KKR.RHS(10, 100, 5, print_output=False))
```
